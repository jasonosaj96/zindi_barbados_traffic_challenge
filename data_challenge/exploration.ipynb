```python
# Load from the video_processed_files directory the jsons and load them as a dataframe
import json
import pandas as pd
import os
data_dir = '../video_processed_files'
all_data = []
for subdir in os.listdir(data_dir):
    subdir_path = os.path.join(data_dir, subdir)
    if os.path.isdir(subdir_path):
        for filename in os.listdir(subdir_path):
            if filename.endswith('.json'):
                with open(os.path.join(subdir_path, filename), 'r') as f:
                    data = json.load(f)
                    # save a column 'videos' which is the folder name/
                    data['videos'] = f"{subdir}/{filename.replace('.json','.mp4')}"
                    all_data.append(data)
segmented_video_df = pd.DataFrame(all_data)
```

```python
# merge the two dataframes on videos column
merged_df = pd.merge(df, segmented_video_df, on='videos', how='inner')
merged_df
```
