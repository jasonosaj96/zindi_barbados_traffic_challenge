```python
from astral import LocationInfo
from astral.sun import sun
from datetime import datetime, time
import pytz

# Barbados location (approximate coordinates for Bridgetown)
barbados = LocationInfo("Barbados", "Caribbean", "America/Barbados", 13.1939, -59.5432)

def get_sun_times(date):
    """Get sunrise and sunset times for a given date in Barbados"""
    s = sun(barbados.observer, date=date, tzinfo=pytz.timezone('America/Barbados'))
    return s['sunrise'].time(), s['sunset'].time()

def classify_time_period(dt):
    """
    Classify time into period of day
    - Night: sunset to sunrise
    - Morning: sunrise to 12:00
    - Afternoon: 12:00 to 17:00
    - Evening: 17:00 to sunset
    """
    if isinstance(dt, str):
        dt = pd.to_datetime(dt)
    
    date = dt.date()
    time_of_day = dt.time()
    
    sunrise, sunset = get_sun_times(date)
    
    # Determine if it's day or night
    if time_of_day < sunrise or time_of_day >= sunset:
        day_night = 'night'
    else:
        day_night = 'day'
    
    # Determine period of day
    if time_of_day < sunrise:
        period = 'night'
    elif sunrise <= time_of_day < time(12, 0):
        period = 'morning'
    elif time(12, 0) <= time_of_day < time(17, 0):
        period = 'afternoon'
    elif time(17, 0) <= time_of_day < sunset:
        period = 'evening'
    else:
        period = 'night'
    
    return day_night, period, sunrise, sunset

# Test with a sample datetime
test_date = datetime(2025, 10, 20, 14, 30)
day_night, period, sunrise, sunset = classify_time_period(test_date)
print(f"Test: {test_date}")
print(f"Sunrise: {sunrise}, Sunset: {sunset}")
print(f"Day/Night: {day_night}, Period: {period}")
```
