```python
features_df = pd.read_csv('../training_dataset.csv')
# columns to predict north_entry_congestion, east_entry_congestion, south_entry_congestion, west_entry_congestion
features_df.head()
```

```python
features_df['video_timestamp'] # parse this column as datetime there is a -45 seconds at the end
features_df['video_timestamp'] = pd.to_datetime(features_df['video_timestamp'], format='%Y-%m-%d-%H-%M-%S')
```

```python
# sort by the video_timestamp
features_df = features_df.sort_values(by=['video_timestamp'])
features_df.head()
```

## Timestamp Analysis

```python
# Basic time range
print("Time Range:")
print(f"Earliest: {features_df['video_timestamp'].min()}")
print(f"Latest: {features_df['video_timestamp'].max()}")
print(f"Duration: {features_df['video_timestamp'].max() - features_df['video_timestamp'].min()}")
```

```python
# Number of unique timestamps and check for gaps
print(f"\nTotal records: {len(features_df)}")
print(f"Unique timestamps: {features_df['video_timestamp'].nunique()}")

# Check the time differences between consecutive timestamps
time_diffs = features_df['video_timestamp'].diff()
print(f"\nTime interval between recordings:")
print(f"Most common interval: {time_diffs.mode()[0]}")
print(f"Min interval: {time_diffs.min()}")
print(f"Max interval: {time_diffs.max()}")
```
